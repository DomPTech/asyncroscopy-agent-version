# AI Agent Demo

Minimal proof of concept that asks a local LLM agent to use tools from the asyncroscopy MCP server.

This notebook talks to **MCP only**. The agent swarm is served by its own MCP server (`run_llm.py`), which inherits the microscope tools from the instrument MCP server (`run_mcp.py`), and acquired data comes back through the `get_data_from_key` tool.

## Prereqs

Start the Tango stack on the microscope computer. Start the instrument MCP server and the
agent MCP server on the LLM computer. Only the instrument MCP server needs the microscope
computer's Tango database:

```bash
uv run startup_scripts/run_servers.py
uv run startup_scripts/run_mcp.py --yaml configs/mcp.yaml
uv run startup_scripts/run_llm.py --yaml configs/gemma-llm.yaml
```

If not already done, install the optional AI extras:

```bash
uv sync --extra agent
# for local models via Ollama do uv sync --extra agent --extra ollama
```

## Imports

````python
import json

import numpy as np
from fastmcp import Client
````

## Connect to the MCP servers

We connect to the agent swarm we prompt, and the instrument server whose tools acquire data and read it back.

````python
AGENT_MCP_URL = "http://localhost:8001/mcp"
INSTRUMENT_MCP_URL = "http://localhost:8000/mcp"

agent = Client(AGENT_MCP_URL)
instrument = Client(INSTRUMENT_MCP_URL)

async with instrument:
    devices = await instrument.call_tool("list_devices", {})

print("Devices:", list(devices.data))
````

## Spawn an agent

`spawn_agent` takes a JSON config. `tools` accepts glob patterns matched against the
inherited MCP tool names, so `*spectrum` binds only the spectrum tools.

````python
eds_agent_config = json.dumps({
    "name": "eds",
    "system_prompt": (
        "You are a worker agent responsible for acquiring EDS spectra."
    ),
    "tools": ["*spectrum"],
    "description": "Agent responsible for acquiring EDS spectra using available tools."
})

async with agent:
    spawned = await agent.call_tool("spawn_agent", {"config": eds_agent_config})
    print(spawned.data)
````

````python
async with agent:
    agents = await agent.call_tool("list_agents", {})
    inherited = await agent.call_tool("list_agent_tools", {})

print("Agents:", list(agents.data))
print("Inherited tools:", len(json.loads(inherited.data)), "e.g.", [t["name"] for t in json.loads(inherited.data)[:5]])
````

## Give a prompt

`query_agent` runs the swarm and returns its final response. If the agent cannot complete
a task, raise the LangGraph step limit with `set_max_steps`.

````python
prompt = "Get a scanned haadf image. Then get an EDS spectrum."

async with agent:
    result = await agent.call_tool("query_agent", {"prompt": prompt})

response = result.data
print(response)
````

## Fetch the acquired data

The Tiled key each acquisition returned is printed above in the agent's tool output.
Copy the image key and the spectrum key here.

````python
image_key = "image_HAADF_20260731T170000000000.h5"
spectrum_key = "spectrum_eds_20260731T165346932622.h5"
````

````python
async def get_data(key, max_values=1_000_000):
    async with Client(INSTRUMENT_MCP_URL) as client:
        return (await client.call_tool("get_data_from_key", {"key": key, "max_values": max_values})).data

recent = {image_key: await get_data(image_key), spectrum_key: await get_data(spectrum_key)}
````

`get_data_from_key` returns each dataset's full `shape`, `dtype`, and `attrs`, plus a `preview` of at most `max_values` leading values. Raise `max_values` to read more of of larger data like an image.

### Plot HAADF Images

````python
import matplotlib.pyplot as plt

def read_preview(dataset):
    """Reshape a dataset preview into the largest block its values actually fill."""
    values = np.asarray(dataset["preview"], dtype=float)
    full = int(np.prod(dataset["shape"]))
    if values.size == full:
        return values.reshape(dataset["shape"])

    width = dataset["shape"][-1]
    if width > values.size:
        print(f"preview holds {values.size} of {dataset['shape']} values; raising max_values")
        return values.reshape(1, -1)

    rows = values.size // width
    block = values[: rows * width].reshape(rows, width)
    print(f"preview is {rows}x{width} of the full {dataset['shape']}")
    return block

dataset = next(d for d in recent[image_key]["datasets"] if d["name"] == "image/HAADF")
print("Metadata:", dataset["attrs"])

image = read_preview(dataset)
print("Array shape:", image.shape)

fig, ax = plt.subplots(figsize=(6, 6))
ax.imshow(image, cmap="gray", interpolation="none")
ax.set_title("Result")
ax.axis("off")
plt.tight_layout()
````

### Plot EDS Spectrum

````python
dataset = next(d for d in recent[spectrum_key]["datasets"] if d["name"].startswith("spectrum"))

spectrum = read_preview(dataset)

print("Tiled key     :", spectrum_key)
print("Metadata      :", dataset["attrs"])
print("Spectrum shape:", spectrum.shape)
spectrum

plt.figure(figsize=(6, 4))
plt.plot(spectrum)
plt.show()
````
